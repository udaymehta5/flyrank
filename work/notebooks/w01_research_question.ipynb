## 1. My Lane

**Lane:** Structured Content Archetype Clustering (Core lane)
**Data:** Warehouse / starter dataset

Grouping pages by structural traits (word count, position tier, content type)
into archetypes — not by topic/meaning. This is structural clustering,
not semantic clustering.

## 2. The Question

**Question:** Do pages cluster into distinct structural archetypes, and do
these archetypes differ in performance?

**Unit of analysis:** A page.

**Output:** Cluster labels + a performance/trait profile per cluster.

**Action:** Prioritize restructuring pages in low-performing archetypes.

**Cost of wrong call:** Wasted effort restructuring pages that didn't need it,
or missing ones that did.

**Why ML:** Archetypes emerge from multiple features interacting — no single
column or rule reveals them; needs unsupervised discovery.

## 3. Quick Look at the Data

```python
import os

for root, dirs, files in os.walk("data"):
    for f in files:
        print(os.path.join(root, f))
```

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print(f"Shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
```

```python
combo_count = df.groupby(["content_type", "position_tier"]).ngroups
total_pages = len(df)
wc_std = df["word_count"].std()
wc_mean = df["word_count"].mean()

print(f"Unique content_type x position_tier combos: {combo_count}")
print(f"Total pages: {total_pages}")
print(f"Word count — mean: {wc_mean:.1f}, std: {wc_std:.1f}")
print(df["content_type"].value_counts())

print("\n--- Copy the line below into Section 3 markdown ---\n")
print(f"{combo_count} combinations across {total_pages} pages, with std of {wc_std:.1f} in word count — enough structural variety to support clustering.")
```

 14 combinations across 30000 pages, with std of 1452.4 in word count — enough structural variety to support clustering.

## 4. Careful Words

- Clusters show correlation, not causation.
- Not semantic — says nothing about content meaning/topic.
- Cluster boundaries may shift with more features or full warehouse data.
